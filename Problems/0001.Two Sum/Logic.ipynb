# LeetCode 1: Two Sum

## Problem Statement

Given an array of integers nums and an integer target,
return the indices of the two numbers such that they add
up to the target.

You may not use the same element twice.

## Approach 1: Brute Force

### Intuition
Check every possible pair of elements using nested loops.

### Algorithm
1. Iterate through the array.
2. Use another loop to check the remaining elements.
3. If their sum equals the target, return their indices.

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        n = len(nums)

        for i in range(n):
            for j in range(i + 1, n):
                if nums[i] + nums[j] == target:
                    return [i, j]


# Test
sol = Solution()
print(sol.twoSum([2, 7, 11, 15], 9))
```

### Complexity Analysis

- Time Complexity: O(n²)
- Space Complexity: O(1)

The nested loops check every possible pair.

## Approach 2: Optimized HashMap Solution

### Intuition

Instead of checking every pair, use a dictionary to
store previously visited numbers and their indices.

For every number, calculate: `complement = target - num`

If the complement already exists in the dictionary,
return both indices.

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        seen = {}
        for i, num in enumerate(nums):
            complement=target-num
            if complement in seen:
                return [seen[complement], i]

            seen[num] = i


# Test
sol = Solution()
print(sol.twoSum([2, 7, 11, 15], 9))
```

### Complexity Analysis

- Time Complexity: `O(n) average`
- Space Complexity: `O(n)`

Dictionary lookups and insertions take `O(1)` average time.

## Final Comparison

| Approach | Time Complexity | Space Complexity |
|---|---|---|
| Brute Force | O(n²) | O(1) |
| HashMap | O(n) average | O(n) |

**Optimal Solution: HashMap – O(n) average time.**
