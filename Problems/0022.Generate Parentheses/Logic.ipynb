# [Generate Parentheses](https://leetcode.com/problems/generate-parentheses/)

## Problem Statement

Given n pairs of parentheses, write a function to generate all combinations of well-formed parentheses.

### Example 1
Input: n = 3

Output:
["((()))", "(()())", "(())()", "()(())", "()()()"]

### Example 2
Input: n = 1

Output: ["()"]

```python
class Solution:
    def generateParenthesis(self, n: int) -> list[str]:
        result = []

        def isvalid(s):
            balance = 0

            for char in s:
                if char == '(': balance += 1
                else: balance -= 1

                if balance < 0: return False
            return balance == 0

        def generate(s):
            if len(s) == 2 * n:
                if isvalid(s):
                    result.append(s)
                return

            generate(s + '(')
            generate(s + ')')

        generate('')
        return result


# Test
sol = Solution()
print(sol.generateParenthesis(3))
```


### Complexity Analysis

- Time: `O(n * 4^n)`
- Auxiliary Space: `O(n)`, excluding output.

**Limitation:** Generates invalid strings and validates them afterward.


```python
class Solution:
    def backtrack(self, current:str, open:int, close:int, n:int):
        # Base case
        if len(current)==2*n:
            self.result.append(current)
            return 
        
        # Add opening bracket
        if open<n:
            self.backtrack(current+'(',open+1,close,n)

        # Add closing bracket
        if close<open:
            self.backtrack(current+')',open,close+1,n)
    
    def generateParenthesis(self, n: int) -> list[str]:
        self.result=[]
        self.backtrack("",0,0,n)

        return self.result

# Test
sol=Solution()
sol.generateParenthesis(3)
```


### Complexity Analysis

Let C(n) be the nth Catalan number.

- Time: `O(n * C(n))`
- Auxiliary Space: O(n), excluding output.
- Output Space: `O(n * C(n))`.

**Why optimal?** Only valid combinations are generated.


```python

class Solution:
    def generateParenthesis(self, n: int) -> list[str]:

        dp = [[] for _ in range(n + 1)]
        dp[0] = ['']

        for i in range(1, n + 1):
            for j in range(i):
                for left in dp[j]:
                    for right in dp[i-1-j]:
                        dp[i].append('(' + left + ')' + right)

        return dp[n]


# Test
sol = Solution()
print(sol.generateParenthesis(3))
```


### Complexity Analysis

- Time: O(n² * C(n)) upper bound.
- Auxiliary Space: O(n * C(n)) for stored combinations.
- Including string storage: O(n² * C(n)).

**Intuition:** Reuse previously generated combinations to build larger solutions.



## Final Comparison

| Approach | Time Complexity | Auxiliary Space |
|---|---|---|
| Brute Force | O(n * 4^n) | O(n) |
| Backtracking | O(n * C(n)) | O(n) |
| Dynamic Programming | O(n² * C(n)) | O(n * C(n)) |


## Interview Recommendation
**Backtracking is the recommended approach.**
It generates only valid combinations and is the most natural solution to explain in an interview.
